```python
df_model = df_long[df_long['year'] >= 1970].copy()
print(f"df_model shape: {df_model.shape}")
print(f"Year range: {df_model['year'].min()}-{df_model['year'].max()}")
print(f"Countries: {df_model['ISO3'].nunique()}")
print(f"Non-missing infl: {df_model['infl'].notna().sum()}")
```

```python
FEATURES = ['infl', 'rGDP_pc', 'unemp', 'govdebt_GDP',
            'CA_GDP', 'USDfx', 'REER', 'pop']

for feat in FEATURES:
    if feat in df_model.columns:
        df_model[feat] = df_model.groupby('ISO3')[feat].ffill(limit=2)
        df_model[feat] = df_model.groupby('ISO3')[feat].bfill(limit=2)

print("After ffill/bfill:")
for feat in FEATURES:
    n = df_model[feat].notna().sum()
    print(f"  {feat:15s} | {n} / {len(df_model)} ({100*n/len(df_model):.1f}%)")
```

```python
# Target at year t = mean(infl[t+1], ..., infl[t+5])
df_model = df_model.sort_values(['ISO3', 'year']).reset_index(drop=True)

df_model['target_5y'] = df_model.groupby('ISO3')['infl'].transform(
    lambda s: s.rolling(window=5, min_periods=5).mean().shift(-5)
)

valid_targets = df_model[df_model['target_5y'].notna()]
print(f"Target valid: {len(valid_targets)} samples")
print(f"Target year range: {valid_targets['year'].min()}-{valid_targets['year'].max()}")

# Per-country target count
per_country = valid_targets.groupby('ISO3').size()
print(f"\nCountries: {len(per_country)}")
print(f"Min/Max/Mean targets per country: {per_country.min()}/{per_country.max()}/{per_country.mean():.1f}")
```

```python
valid = df_model['target_5y'].dropna()

print("=== Target distribution ===")
print(valid.describe())

# Clip 5-95%
t_lower = valid.quantile(0.05)
t_upper = valid.quantile(0.95)
print(f"\nClipping to [{t_lower:.2f}, {t_upper:.2f}]")

df_model['target_clipped'] = df_model['target_5y'].clip(t_lower, t_upper)

# Log transform
shift = -df_model['target_clipped'].min() + 1
df_model['target_log'] = np.log(df_model['target_clipped'] + shift)
print(f"Shift: {shift:.4f}")

# Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].hist(df_model['target_clipped'].dropna(), bins=50, edgecolor='black')
axes[0].set_title('Target (clipped, %)')
axes[0].set_xlabel('5-year avg inflation')

axes[1].hist(df_model['target_log'].dropna(), bins=50, edgecolor='black', color='orange')
axes[1].set_title('Target (log-transformed)')
axes[1].set_xlabel('log(5-year avg)')
plt.tight_layout()
plt.show()
```

```python
WINDOW = 10
FEATURES = ['infl', 'rGDP_pc', 'unemp', 'govdebt_GDP',
            'CA_GDP', 'USDfx', 'REER', 'pop']

# Scaler (fit on train only later)
from sklearn.preprocessing import RobustScaler, StandardScaler

# First scale features (will re-fit on train later)
scaler_X = RobustScaler()
scaler_y = StandardScaler()

# Use all valid rows for now
valid_mask = df_model['target_log'].notna()
scaler_X.fit(df_model.loc[valid_mask, FEATURES])
scaler_y.fit(df_model.loc[valid_mask, ['target_log']])

df_model.loc[valid_mask, FEATURES] = scaler_X.transform(
    df_model.loc[valid_mask, FEATURES]
)
df_model.loc[valid_mask, 'target_scaled'] = scaler_y.transform(
    df_model.loc[valid_mask, ['target_log']]
)

print("Scaling done.")
print(f"Scaler_y mean={scaler_y.mean_[0]:.4f}, scale={scaler_y.scale_[0]:.4f}")
```

```python
def make_windows(df, features, window_size):
    X_list, y_list, meta = [], [], []
    for iso, group in df.groupby('ISO3'):
        group = group.sort_values('year').reset_index(drop=True)
        n = len(group)
        for i in range(window_size, n):
            window = group.iloc[i-window_size:i]
            target_row = group.iloc[i]
            if pd.isna(target_row['target_scaled']):
                continue
            if window[features].isna().any().any():
                continue
            X_list.append(window[features].values)
            y_list.append(target_row['target_scaled'])
            meta.append({
                'ISO3': iso,
                'pred_year': int(target_row['year']),
                'target_real': target_row['target_clipped']
            })
    return np.array(X_list, dtype=np.float32), np.array(y_list, dtype=np.float32), pd.DataFrame(meta)

X_all, y_all, meta_all = make_windows(df_model, FEATURES, WINDOW)

print(f"X_all: {X_all.shape}, y_all: {y_all.shape}")
print(f"Pred year range: {meta_all['pred_year'].min()}-{meta_all['pred_year'].max()}")
print(f"\nPred year distribution (by 5-year bins):")
meta_all['bin'] = (meta_all['pred_year'] // 5) * 5
print(meta_all.groupby('bin').size())
```

```python
# Split: Train ≤ 2000, Gap 2001-2005, Test 2006-2015
# এই split-এ test-এ realized data guaranteed

TRAIN_END = 2000    # target covers 2001-2005
TEST_START = 2006   # window 1996-2006
TEST_END = 2015     # target covers 2016-2020

train_idx = meta_all['pred_year'] <= TRAIN_END
test_idx = (meta_all['pred_year'] >= TEST_START) & (meta_all['pred_year'] <= TEST_END)

print(f"Train: {train_idx.sum()} samples")
print(f"Test:  {test_idx.sum()} samples")
print(f"\nTrain pred_year: {meta_all[train_idx]['pred_year'].min()}-{meta_all[train_idx]['pred_year'].max()}")
print(f"Test pred_year:  {meta_all[test_idx]['pred_year'].min()}-{meta_all[test_idx]['pred_year'].max()}")

# Gap check
train_target_max = meta_all[train_idx]['pred_year'].max() + 5
test_target_min = meta_all[test_idx]['pred_year'].min() + 1
print(f"\nTrain targets end: {train_target_max}")
print(f"Test targets start: {test_target_min}")
print(f"Gap: {test_target_min - train_target_max - 1} years")

X_train, y_train = X_all[train_idx], y_all[train_idx]
X_test, y_test = X_all[test_idx], y_all[test_idx]
meta_train = meta_all[train_idx].reset_index(drop=True)
meta_test = meta_all[test_idx].reset_index(drop=True)
```

```python
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
import xgboost as xgb
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

X_train_flat = X_train.reshape(X_train.shape[0], -1)
X_test_flat = X_test.reshape(X_test.shape[0], -1)

def inverse_pipeline(pred):
    log_p = scaler_y.inverse_transform(pred.reshape(-1, 1)).flatten()
    return np.exp(log_p) - shift

y_train_real = inverse_pipeline(y_train)
y_test_real = inverse_pipeline(y_test)

results = {}

# 1. Naive
results['Naive Mean'] = {
    'rmse': np.sqrt(mean_squared_error(y_test_real, np.full_like(y_test_real, y_train_real.mean()))),
    'mae': mean_absolute_error(y_test_real, np.full_like(y_test_real, y_train_real.mean())),
    'r2': r2_score(y_test_real, np.full_like(y_test_real, y_train_real.mean()))
}

# 2. Persistence
persistence_preds = []
for _, row in meta_test.iterrows():
    iso = row['ISO3']
    py = row['pred_year']
    prev = df_long[
        (df_long['ISO3'] == iso) &
        (df_long['year'] >= py - 5) &
        (df_long['year'] < py)
    ]['infl'].mean()
    persistence_preds.append(prev)
persistence_preds = np.array(persistence_preds)
persistence_preds = np.clip(persistence_preds, t_lower, t_upper)

results['Persistence'] = {
    'rmse': np.sqrt(mean_squared_error(y_test_real, persistence_preds)),
    'mae': mean_absolute_error(y_test_real, persistence_preds),
    'r2': r2_score(y_test_real, persistence_preds)
}

# 3-6. Models
for name, model in [
    ('Ridge', Ridge(alpha=100)),
    ('Random Forest', RandomForestRegressor(n_estimators=200, max_depth=10, random_state=42, n_jobs=-1)),
    ('Gradient Boosting', GradientBoostingRegressor(n_estimators=200, max_depth=3, learning_rate=0.05, random_state=42)),
    ('XGBoost', xgb.XGBRegressor(n_estimators=200, max_depth=3, learning_rate=0.1, random_state=42, n_jobs=-1))
]:
    model.fit(X_train_flat, y_train)
    pred = model.predict(X_test_flat)
    pred_real = inverse_pipeline(pred)
    results[name] = {
        'rmse': np.sqrt(mean_squared_error(y_test_real, pred_real)),
        'mae': mean_absolute_error(y_test_real, pred_real),
        'r2': r2_score(y_test_real, pred_real)
    }
    print(f"{name}: R²={results[name]['r2']:.4f}")

# Summary
summary = pd.DataFrame(results).T.round(4).sort_values('r2', ascending=False)
print("\n" + "="*70)
print(f"RESULTS (Window={WINDOW}, Test {TEST_START}-{TEST_END})")
print("="*70)
print(summary.to_string())
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import TensorDataset, DataLoader

# Model architecture
class TemporalAttention(nn.Module):
    def __init__(self, h):
        super().__init__()
        self.attn = nn.Linear(h, 1, bias=False)
    def forward(self, lstm_out):
        w = F.softmax(self.attn(lstm_out), dim=1)
        return torch.sum(w * lstm_out, dim=1), w

class CNN_LSTM_Attention(nn.Module):
    def __init__(self, n_features, seq_len, cnn_channels=(32, 64),
                 lstm_hidden=64, lstm_layers=2, dropout=0.3):
        super().__init__()
        self.conv1 = nn.Conv1d(n_features, cnn_channels[0], 3, padding=1)
        self.bn1 = nn.BatchNorm1d(cnn_channels[0])
        self.conv2 = nn.Conv1d(cnn_channels[0], cnn_channels[1], 3, padding=1)
        self.bn2 = nn.BatchNorm1d(cnn_channels[1])
        self.dropout = nn.Dropout(dropout)
        self.lstm = nn.LSTM(cnn_channels[1], lstm_hidden, lstm_layers,
                            batch_first=True, dropout=dropout if lstm_layers > 1 else 0)
        self.attn = TemporalAttention(lstm_hidden)
        self.fc = nn.Sequential(
            nn.Linear(lstm_hidden, 32), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(32, 1)
        )
    def forward(self, x):
        x = x.transpose(1, 2)
        x = F.relu(self.bn1(self.conv1(x)))
        x = F.relu(self.bn2(self.conv2(x)))
        x = self.dropout(x)
        x = x.transpose(1, 2)
        lstm_out, _ = self.lstm(x)
        context, _ = self.attn(lstm_out)
        return self.fc(context).squeeze(-1), _

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
torch.manual_seed(42)

model = CNN_LSTM_Attention(X_train.shape[2], X_train.shape[1]).to(device)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=5e-4, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=10)

train_ds = TensorDataset(torch.tensor(X_train), torch.tensor(y_train))
test_ds = TensorDataset(torch.tensor(X_test), torch.tensor(y_test))
train_loader = DataLoader(train_ds, batch_size=32, shuffle=True)
test_loader = DataLoader(test_ds, batch_size=32, shuffle=False)

def train_epoch(m, loader, opt, crit):
    m.train()
    total, n = 0, 0
    for X_b, y_b in loader:
        X_b, y_b = X_b.to(device), y_b.to(device)
        opt.zero_grad()
        pred, _ = m(X_b)
        loss = crit(pred, y_b)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(m.parameters(), 1.0)
        opt.step()
        total += loss.item() * X_b.size(0)
        n += X_b.size(0)
    return total / n

@torch.no_grad()
def eval_m(m, loader, crit):
    m.eval()
    total, n = 0, 0
    preds = []
    for X_b, y_b in loader:
        X_b, y_b = X_b.to(device), y_b.to(device)
        pred, _ = m(X_b)
        loss = crit(pred, y_b)
        total += loss.item() * X_b.size(0)
        n += X_b.size(0)
        preds.append(pred.cpu().numpy())
    return total / n, np.concatenate(preds)

best_loss = float('inf')
patience_cnt = 0
best_state = None

for epoch in range(1, 201):
    tl = train_epoch(model, train_loader, optimizer, criterion)
    vl, _ = eval_m(model, test_loader, criterion)
    scheduler.step(vl)
    if vl < best_loss:
        best_loss = vl
        patience_cnt = 0
        best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
    else:
        patience_cnt += 1
    if epoch % 20 == 0:
        print(f"Epoch {epoch:3d} | Train: {tl:.4f} | Val: {vl:.4f}")
    if patience_cnt >= 20:
        print(f"Early stop at epoch {epoch}")
        break

model.load_state_dict(best_state)
_, pred_cnn = eval_m(model, test_loader, criterion)
pred_cnn_real = inverse_pipeline(pred_cnn)

results['CNN-LSTM'] = {
    'rmse': np.sqrt(mean_squared_error(y_test_real, pred_cnn_real)),
    'mae': mean_absolute_error(y_test_real, pred_cnn_real),
    'r2': r2_score(y_test_real, pred_cnn_real)
}

print("\n" + "="*70)
print("FINAL RESULTS")
print("="*70)
final_df = pd.DataFrame(results).T.round(4).sort_values('r2', ascending=False)
print(final_df.to_string())
```

```python
# Second fold: Train ≤ 2005, Test 2016-2020
TRAIN_END_2 = 2005
TEST_START_2 = 2016
TEST_END_2 = 2020

train_idx_2 = meta_all['pred_year'] <= TRAIN_END_2
test_idx_2 = (meta_all['pred_year'] >= TEST_START_2) & (meta_all['pred_year'] <= TEST_END_2)

X_train_2, y_train_2 = X_all[train_idx_2], y_all[train_idx_2]
X_test_2, y_test_2 = X_all[test_idx_2], y_all[test_idx_2]
meta_train_2 = meta_all[train_idx_2].reset_index(drop=True)
meta_test_2 = meta_all[test_idx_2].reset_index(drop=True)

print(f"Fold 2 - Train: {train_idx_2.sum()}, Test: {test_idx_2.sum()}")
print(f"Train pred_year: {meta_train_2['pred_year'].min()}-{meta_train_2['pred_year'].max()}")
print(f"Test pred_year: {meta_test_2['pred_year'].min()}-{meta_test_2['pred_year'].max()}")
```

```python
X_train_2_flat = X_train_2.reshape(X_train_2.shape[0], -1)
X_test_2_flat = X_test_2.reshape(X_test_2.shape[0], -1)

y_train_2_real = inverse_pipeline(y_train_2)
y_test_2_real = inverse_pipeline(y_test_2)

results_2 = {}

# Naive
results_2['Naive'] = {
    'r2': r2_score(y_test_2_real, np.full_like(y_test_2_real, y_train_2_real.mean()))
}

# Persistence
persistence_2 = []
for _, row in meta_test_2.iterrows():
    iso = row['ISO3']
    py = row['pred_year']
    prev = df_long[
        (df_long['ISO3'] == iso) &
        (df_long['year'] >= py - 5) &
        (df_long['year'] < py)
    ]['infl'].mean()
    persistence_2.append(prev)
persistence_2 = np.array(persistence_2)
persistence_2 = np.clip(persistence_2, t_lower, t_upper)

results_2['Persistence'] = {
    'r2': r2_score(y_test_2_real, persistence_2)
}

# Tree models
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
import xgboost as xgb

for name, model in [
    ('Random Forest', RandomForestRegressor(n_estimators=200, max_depth=10, random_state=42, n_jobs=-1)),
    ('XGBoost', xgb.XGBRegressor(n_estimators=200, max_depth=3, learning_rate=0.1, random_state=42, n_jobs=-1)),
    ('Gradient Boosting', GradientBoostingRegressor(n_estimators=200, max_depth=3, learning_rate=0.05, random_state=42))
]:
    model.fit(X_train_2_flat, y_train_2)
    pred = model.predict(X_test_2_flat)
    pred_real = inverse_pipeline(pred)
    results_2[name] = {
        'r2': r2_score(y_test_2_real, pred_real),
        'rmse': np.sqrt(mean_squared_error(y_test_2_real, pred_real))
    }

print("\n=== Fold 2 Results (Test 2016-2020) ===")
for name, vals in results_2.items():
    print(f"  {name:20s} | R²={vals['r2']:+.4f}")
```

```python
# CNN-LSTM on Fold 2
torch.manual_seed(42)
model_f2 = CNN_LSTM_Attention(X_train_2.shape[2], X_train_2.shape[1]).to(device)
criterion_f2 = nn.MSELoss()
opt_f2 = torch.optim.Adam(model_f2.parameters(), lr=5e-4, weight_decay=1e-4)
sched_f2 = torch.optim.lr_scheduler.ReduceLROnPlateau(opt_f2, mode='min', factor=0.5, patience=10)

train_ds_f2 = TensorDataset(torch.tensor(X_train_2), torch.tensor(y_train_2))
test_ds_f2 = TensorDataset(torch.tensor(X_test_2), torch.tensor(y_test_2))
train_loader_f2 = DataLoader(train_ds_f2, batch_size=32, shuffle=True)
test_loader_f2 = DataLoader(test_ds_f2, batch_size=32, shuffle=False)

best_loss_f2 = float('inf')
patience_f2 = 0
best_state_f2 = None

for epoch in range(1, 201):
    tl = train_epoch(model_f2, train_loader_f2, opt_f2, criterion_f2)
    vl, _ = eval_m(model_f2, test_loader_f2, criterion_f2)
    sched_f2.step(vl)
    if vl < best_loss_f2:
        best_loss_f2 = vl
        patience_f2 = 0
        best_state_f2 = {k: v.cpu().clone() for k, v in model_f2.state_dict().items()}
    else:
        patience_f2 += 1
    if patience_f2 >= 20:
        print(f"Early stop at epoch {epoch}")
        break

model_f2.load_state_dict(best_state_f2)
_, pred_cnn_2 = eval_m(model_f2, test_loader_f2, criterion_f2)
pred_cnn_2_real = inverse_pipeline(pred_cnn_2)

r2_cnn_f2 = r2_score(y_test_2_real, pred_cnn_2_real)
print(f"CNN-LSTM Fold 2 R²: {r2_cnn_f2:+.4f}")
```
